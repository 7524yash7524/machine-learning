# **What Is Machine Learning?**

**Machine learning is the study and development of algorithms that allow computers to learn from data and make predictions or decisions based on that learning.**

**Simple explanation**

*Instead of writing specific rules for every situation, you provide the computer with examples (data). The machine learning algorithm analyzes those examples, identifies patterns, and uses those patterns to make predictions or decisions about new, unseen data.*

**Example**

*Suppose you want a computer to recognize spam emails:*

**Traditional programming:** *You manually write rules like "if the email contains certain words, mark it as spam."*

**Machine learning:** *You give the computer thousands of emails labeled as "spam" or "not spam." It learns the patterns that distinguish spam from legitimate emails and can classify new emails on its own.*

# **Why Use Machine Learning?**

*We use machine learning when it is difficult or impossible to write explicit rules for solving a problem.Instead of programming every rule manually, we let the computer learn those rules from data.*

**Example :** *Spam Email Detection (Traditional Programming vs Machine Learning)*

Imagine you want to detect spam emails.

Traditional Programming

You write rules like:

IF email contains "Win Money"
AND sender is unknown
THEN Spam

You continue adding rules:

If it contains "Congratulations"
If it has too many links
If it has strange characters
If it has many capital letters

Eventually you'll have thousands of rules, and spammers will simply change their wording.

This approach becomes difficult to maintain.

**Machine Learning**

Instead, collect data like this:

Email	Spam?
Win an iPhone today	Yes
Meeting at 3 PM	No
Claim your prize	Yes
Project update	No

The algorithm studies thousands of examples and automatically learns patterns that distinguish spam from legitimate emails.

When a new email arrives:

New Email
      ↓
Machine Learning Model
      ↓
Spam / Not Spam

No one had to manually write every rule.

# **Types of Machine Learning Systems**

**There are so many different types of machine learning systems that it is useful to classify them in broad categories, based on the following criteria:**

*How they are guided during training (supervised,unsupervised, semi-supervised, self-supervised, and others)
Whether or not they can learn incrementally on the fly (online versus batch learning) Whether they work by simply comparing new data points to known data points, or instead by detecting patterns in the training data and building a predictive model, much like scientists do (instance-based versus model-based learning) These criteria are not exclusive; you can combine them in any way you like. For example, a state-of-the-art spam filter may learn on the fly using a deep neural network model trained using human-provided examples of spam and ham; this makes it an online, model-based, supervised learning system.*

Let’s look at each of these criteria a bit more closely

## **Supervised learning**

**In supervised learning, the training set you feed to the algorithm
includes the desired solutions, called labels**

## Definition

**Supervised learning** is a type of machine learning where the model learns from **labeled data**. Labeled data means that each input has a corresponding **correct output (label)**. The goal is to learn the relationship between inputs and outputs so that the model can accurately predict the output for new, unseen data.

> **Definition (Interview):**
> Supervised learning is a type of machine learning in which a model learns from labeled data (inputs with known outputs) to predict the correct output for new, unseen data.

---

# Simple Explanation

Imagine a teacher teaching a student.

The teacher gives:
- **Questions**
- **Correct Answers**

The student studies many examples and eventually learns how to answer new questions without being told the answer.

Similarly, in supervised learning:

- **Teacher** → Training Data
- **Questions** → Input (Features)
- **Correct Answers** → Labels (Target)
- **Student** → Machine Learning Model

---

# Example 1: House Price Prediction

Suppose we have the following dataset:

| House Size (sq ft) | Bedrooms | Price ($) |
|--------------------|----------|-----------|
| 1000 | 2 | 150,000 |
| 1500 | 3 | 220,000 |
| 2000 | 4 | 300,000 |
| 2500 | 4 | 380,000 |

### Step 1: Train the Model

The model receives:

**Inputs (Features):**
- House Size
- Number of Bedrooms

**Output (Label):**
- House Price

It learns the relationship between the features and the price.

For example:

- Larger houses generally cost more.
- Houses with more bedrooms usually have higher prices.

---

### Step 2: Prediction

Now a new house appears:

| House Size | Bedrooms |
|------------|----------|
| 1800 | 3 |

The trained model predicts:

**Predicted Price = \$270,000**

The model has never seen this house before, but it uses what it learned from the training data to estimate its price.

|--------|-------|

| Meeting at 4 PM | Not Spam |

| Project report attached | Not Spam |

The model learns patterns such as:

- Certain words appear frequently in spam emails.

The model predicts:

**Spam**

---

# Why is it Called "Supervised"?

It is called **supervised** because the model is trained using the **correct answers**.

During training:

1. The model makes a prediction.
2. It compares the prediction with the correct answer.
3. It calculates the error.
4. It adjusts itself to reduce that error.

This process repeats many times until the model becomes accurate.

---

# Workflow of Supervised Learning

```text
Training Data
(Input + Correct Output)
        │
        ▼
Machine Learning Algorithm
        │
        ▼
Learns Patterns
        │
        ▼
Trained Model
        │
        ▼
New Input
        │
        ▼
Prediction
```

---

# Types of Supervised Learning

Supervised learning is divided into **two main categories**.

## 1. Regression

Regression is used when the output is a **continuous numerical value**.

### Examples

- Predicting house prices
- Predicting stock prices
- Predicting rainfall
- Predicting temperature

Example:

Input:
- House Size = 1800 sq ft

Output:
- Price = \$270,000

Since the output is a number, this is a **Regression** problem.

---

## 2. Classification

Classification is used when the output belongs to a **category or class**.

### Examples

- Spam or Not Spam
- Cat or Dog
- Disease or Healthy
- Fraud or Genuine
- Positive or Negative Review

Example:

Input:
- Email

Output:
- Spam

Since the output is a category, this is a **Classification** problem.

# Advantages

- High accuracy when enough labeled data is available.
- Easy to evaluate because the correct answers are known.
- Works well for prediction problems.
- Widely used in real-world applications.

---

# Disadvantages

- Requires a large amount of labeled data.
- Labeling data can be expensive and time-consuming.
- Performance depends heavily on the quality of the training data.
- May perform poorly on data that is very different from what it learned.

# Key Points to Remember

## **Unsupervised Learning**

## Definition

**Unsupervised learning** is a type of machine learning where the model learns from **unlabeled data**. Unlike supervised learning, the training data does **not** contain the correct output (label). The goal is to discover hidden patterns, relationships, or structures within the data.

> **Definition (Interview):**
> Unsupervised learning is a type of machine learning in which a model learns from unlabeled data to discover hidden patterns, structures, or relationships without being provided with the correct outputs.

---

# Simple Explanation

Imagine giving a child a box full of different toys without telling them anything about them.

The child starts grouping them naturally:

- Cars together
- Dolls together
- Balls together
- Animals together

No one told the child how to group the toys.

The child **found the patterns by themselves**.

That is exactly how **unsupervised learning** works.

---

# Supervised vs Unsupervised Learning

| Supervised Learning | Unsupervised Learning |
|----------------------|----------------------|
| Uses labeled data | Uses unlabeled data |
| Correct answers are available | No correct answers are available |
| Learns to predict outputs | Learns hidden patterns |
| Teacher provides answers | No teacher is available |

---

# Example 1: Customer Segmentation

No one told the computer these groups existed.

The model discovered them on its own.

# Main Types of Unsupervised Learning

## 1. Clustering

Clustering groups similar data points together.

Example:

Customer data:

```text
Customer A
Customer B
Customer C
Customer D
```

The algorithm forms:

```text
Cluster 1:
A, B

Cluster 2:
C, D
```

### Popular Clustering Algorithms

- K-Means Clustering
- Hierarchical Clustering
- DBSCAN

### Applications

- Customer segmentation
- Image segmentation
- Friend recommendations
- Market analysis
- Document grouping

---

## 2. Association

Association learning finds relationships between items that frequently occur together.

### Example

In a supermarket, the algorithm notices:

People who buy:

- Bread
- Butter

often also buy:

- Milk

This rule can be written as:

```text
Bread + Butter → Milk
```

This is called an **association rule**.

### Applications

- Product recommendations
- Market basket analysis
- Cross-selling products
- Online shopping recommendations

Popular algorithms include:

- Apriori
- FP-Growth

---

## 3. Dimensionality Reduction

Sometimes a dataset has hundreds or even thousands of features.

Many of these features may contain similar information.

Dimensionality reduction reduces the number of features while preserving most of the important information.

### Example

Suppose each student has:

- Height
- Weight
- Age
- Income
- Family Size
- Attendance
- Marks
- Sports Score
- Health Score
- ...

Instead of using all features, the algorithm creates a smaller set of meaningful features.

This makes training:

- Faster
- Simpler
- More efficient

### Popular Algorithms

- Principal Component Analysis (PCA)
- t-SNE
- UMAP
