# **What Is Machine Learning?**

**Machine learning is the study and development of algorithms that allow computers to learn from data and make predictions or decisions based on that learning.**

**Simple explanation**

*Instead of writing specific rules for every situation, you provide the computer with examples (data). The machine learning algorithm analyzes those examples, identifies patterns, and uses those patterns to make predictions or decisions about new, unseen data.*

**Example**

*Suppose you want a computer to recognize spam emails:*

**Traditional programming:** *You manually write rules like "if the email contains certain words, mark it as spam."*

**Machine learning:** *You give the computer thousands of emails labeled as "spam" or "not spam." It learns the patterns that distinguish spam from legitimate emails and can classify new emails on its own.*

# **Why Use Machine Learning?**

*We use machine learning when it is difficult or impossible to write explicit rules for solving a problem.Instead of programming every rule manually, we let the computer learn those rules from data.*

**Example :** *Spam Email Detection (Traditional Programming vs Machine Learning)*

Imagine you want to detect spam emails.

Traditional Programming

You write rules like:

IF email contains "Win Money"
AND sender is unknown
THEN Spam

You continue adding rules:

If it contains "Congratulations"
If it has too many links
If it has strange characters
If it has many capital letters

Eventually you'll have thousands of rules, and spammers will simply change their wording.

This approach becomes difficult to maintain.

**Machine Learning**

Instead, collect data like this:

Email	Spam?
Win an iPhone today	Yes
Meeting at 3 PM	No
Claim your prize	Yes
Project update	No

The algorithm studies thousands of examples and automatically learns patterns that distinguish spam from legitimate emails.

When a new email arrives:

New Email
      ↓
Machine Learning Model
      ↓
Spam / Not Spam

No one had to manually write every rule.

# **Types of Machine Learning Systems**

**There are so many different types of machine learning systems that it is useful to classify them in broad categories, based on the following criteria:**

*How they are guided during training (supervised,unsupervised, semi-supervised, self-supervised, and others)
Whether or not they can learn incrementally on the fly (online versus batch learning) Whether they work by simply comparing new data points to known data points, or instead by detecting patterns in the training data and building a predictive model, much like scientists do (instance-based versus model-based learning) These criteria are not exclusive; you can combine them in any way you like. For example, a state-of-the-art spam filter may learn on the fly using a deep neural network model trained using human-provided examples of spam and ham; this makes it an online, model-based, supervised learning system.*

Let’s look at each of these criteria a bit more closely

## **Supervised learning**

**In supervised learning, the training set you feed to the algorithm
includes the desired solutions, called labels**

## Definition

**Supervised learning** is a type of machine learning where the model learns from **labeled data**. Labeled data means that each input has a corresponding **correct output (label)**. The goal is to learn the relationship between inputs and outputs so that the model can accurately predict the output for new, unseen data.

> **Definition (Interview):**
> Supervised learning is a type of machine learning in which a model learns from labeled data (inputs with known outputs) to predict the correct output for new, unseen data.

---

# Simple Explanation

Imagine a teacher teaching a student.

The teacher gives:
- **Questions**
- **Correct Answers**

The student studies many examples and eventually learns how to answer new questions without being told the answer.

Similarly, in supervised learning:

- **Teacher** → Training Data
- **Questions** → Input (Features)
- **Correct Answers** → Labels (Target)
- **Student** → Machine Learning Model

# Why is it Called "Supervised"?

It is called **supervised** because the model is trained using the **correct answers**.

During training:

1. The model makes a prediction.
2. It compares the prediction with the correct answer.
3. It calculates the error.
4. It adjusts itself to reduce that error.

This process repeats many times until the model becomes accurate.

## **Unsupervised Learning**

## Definition

**Unsupervised learning** is a type of machine learning where the model learns from **unlabeled data**. Unlike supervised learning, the training data does **not** contain the correct output (label). The goal is to discover hidden patterns, relationships, or structures within the data.

> **Definition (Interview):**
> Unsupervised learning is a type of machine learning in which a model learns from unlabeled data to discover hidden patterns, structures, or relationships without being provided with the correct outputs.

---

# Simple Explanation

Imagine giving a child a box full of different toys without telling them anything about them.

The child starts grouping them naturally:

- Cars together
- Dolls together
- Balls together
- Animals together

No one told the child how to group the toys.

The child **found the patterns by themselves**.

That is exactly how **unsupervised learning** works.

This makes training:

- Faster
- Simpler
- More efficient

### Popular Algorithms

- Principal Component Analysis (PCA)
- t-SNE
- UMAP

# Semi-Supervised Learning

## Definition

**Semi-supervised learning** is a type of machine learning that uses **both labeled and unlabeled data** for training. Typically, only a **small portion of the data is labeled**, while the majority is unlabeled. The model uses the labeled data to learn initially and then leverages the unlabeled data to improve its performance.

> **Definition (Interview):**
> Semi-supervised learning is a machine learning technique that trains a model using a small amount of labeled data along with a large amount of unlabeled data to improve prediction accuracy.

---

# Simple Explanation

Imagine a teacher gives you:

- 10 questions with correct answers ✅
- 990 questions without answers ❓

You first learn from the answered questions.

Then, using what you've learned, you try to understand and solve the remaining questions.

This is exactly how **semi-supervised learning** works.

---

# Why Do We Need Semi-Supervised Learning?

In many real-world problems:

- Collecting data is **easy**.
- Labeling data is **expensive, time-consuming, and requires experts**.

For example:

Suppose a hospital has:

- 100,000 X-ray images
- Only 500 images have been labeled by doctors.

Instead of using only the 500 labeled images, semi-supervised learning uses:

- 500 labeled images ✅
- 99,500 unlabeled images ❓

to build a better model.

# **Reinforcement Learning**

## Definition

**Reinforcement Learning (RL)** is a type of machine learning where an **agent learns by interacting with an environment**. Instead of being given correct answers, the agent learns through **trial and error**, receiving **rewards** for good actions and **penalties** for bad actions. The objective is to learn a strategy (called a **policy**) that maximizes the total reward over time.

> **Definition (Interview):**
> Reinforcement learning is a machine learning technique in which an agent learns the best actions by interacting with an environment and receiving rewards or penalties based on its actions.

---

# Simple Explanation

Imagine teaching a dog a new trick.

When the dog performs the correct action, you give it a treat.

When it performs the wrong action, you do not reward it.

Over time, the dog learns which actions earn rewards and starts repeating those actions.

This is exactly how **reinforcement learning** works.

---

# Why Do We Need Reinforcement Learning?

Some problems cannot be solved by simply learning from labeled data because:

- There are no correct answers available beforehand.
- The best action depends on the current situation.
- The agent must make a sequence of decisions.
- The consequences of an action may appear much later.

Examples include:

- Self-driving cars
- Robot navigation
- Playing games like Chess or Go
- Traffic signal control
- Stock trading
- Recommendation systems

---

# Example 1: Teaching a Robot to Walk

Suppose you want a robot to learn how to walk.

Initially, the robot knows nothing.

It starts by trying different actions:

- Move left leg
- Move right leg
- Lean forward
- Balance itself

Some actions make it fall.

Some actions help it move forward.

The environment provides feedback:

| Action | Reward |
|---------|--------|
| Moves forward | +10 |
| Maintains balance | +5 |
| Falls down | -20 |
| Wastes energy | -2 |

After many attempts, the robot learns how to walk without falling.

# Why is it Called "Reinforcement"?

It is called **reinforcement learning** because good behavior is **reinforced** through rewards.

Think of a child learning to ride a bicycle.

- If the child pedals correctly, they move forward.
- If they lose balance, they fall.

Over time, the child learns the best way to ride by repeating successful actions.

Similarly, the agent strengthens actions that lead to higher rewards.

---

# Key Components of Reinforcement Learning

## 1. Agent

The learner or decision-maker.

Examples:

- Robot
- Self-driving car
- Chess-playing computer

---

## 2. Environment

Everything the agent interacts with.

Examples:

- Chess board
- Road
- Video game
- Robot's surroundings

---

## 3. State (S)

The current situation of the agent.

Examples:

- Position of chess pieces
- Current traffic condition
- Robot's location

---

## 4. Action (A)

A decision the agent can make.

Examples:

- Move left
- Accelerate
- Brake
- Pick up an object

---

## 5. Reward (R)

Feedback from the environment.

Examples:

- +10 for reaching a goal
- -20 for crashing
- +1 for collecting a coin

The reward tells the agent whether its action was good or bad.

---

## 6. Policy (π)

A **policy** is the strategy that tells the agent what action to take in each state.

In simple words:

> **Policy = Rule for choosing actions**

The goal of reinforcement learning is to learn the **best policy**.

---

# Workflow of Reinforcement Learning

```text
            +----------------+
            |   Environment  |
            +----------------+
                    ▲
          Reward    │    State
                    │
                    │
            +----------------+
            |     Agent      |
            +----------------+
                    │
                    ▼
                 Action
```

The cycle continues:

1. Agent observes the current state.
2. Agent chooses an action.
3. Environment responds.
4. Agent receives a reward.
5. Agent updates its strategy.
6. Repeat.

Using the best strategy already known.

Taking your usual fastest route to school.

A good reinforcement learning algorithm balances both.

# Batch Learning

## Definition

**Batch Learning** is a learning method where the model is trained using the **entire training dataset at once**. Once the training is complete, the model is deployed and **does not learn from new data automatically**. To learn new information, the model must be **retrained from scratch** using both the old and new data.

> **Definition (Interview):**
> Batch learning is a machine learning approach in which the model is trained on the entire dataset at once and must be retrained whenever new data becomes available.

---

# Simple Explanation

Imagine studying for an exam using an entire textbook.

You finish studying, take the exam, and stop learning.

Later, if a new chapter is added, you must study the **entire book again**.

This is how **batch learning** works.

---

# Workflow

```text
Training Data
      │
      ▼
Train Model
      │
      ▼
Deploy Model
      │
      ▼
New Data Arrives
      │
      ▼
Retrain Entire Model
```

---

# Example 1: House Price Prediction

A company has data for **100,000 houses**.

The model is trained using all of this data.

After one month, 5,000 new house records are collected.

The existing model **cannot learn automatically**.

The company retrains the model using:

- Old data (100,000 houses)
- New data (5,000 houses)

---

# Example 2: Student Grade Prediction

A university trains a model using students' data from the last 10 years.

Next year, thousands of new student records become available.

The university retrains the entire model so it can learn from the new data.

---

# Advantages

- Usually produces highly accurate models.
- Easy to train and evaluate.
- Suitable for datasets that do not change frequently.

---

# Disadvantages

- Retraining can be slow and expensive.
- Cannot immediately learn from new data.
- Requires storing the complete dataset.

---

# Online Learning

## Definition

**Online Learning** is a learning method where the model learns **continuously** by updating itself whenever new data arrives. Instead of retraining from scratch, the model makes **small updates** using the new data.

> **Definition (Interview):**
> Online learning is a machine learning approach in which the model continuously updates itself as new data becomes available without retraining on the entire dataset.

---

# Simple Explanation

Imagine learning one new lesson every day.

You don't study the entire textbook again.

Instead, you simply learn today's lesson and improve your knowledge.

This is how **online learning** works.

---

# Workflow

```text
Initial Training
      │
      ▼
Deploy Model
      │
      ▼
New Data Arrives
      │
      ▼
Update Model
      │
      ▼
Repeat Continuously
```

---

# Example 1: Email Spam Detection

Every day, thousands of new spam emails appear.

Instead of retraining the model from scratch every day, the spam filter updates itself using the latest emails.

This allows it to quickly recognize new spam patterns.

---

# Example 2: Stock Price Prediction

A stock market model receives new price data every second.

Rather than rebuilding the model from the beginning, it continuously updates itself with the latest market information.

This helps it adapt to changing market conditions.

---

# Advantages

- Learns continuously from new data.
- No need to retrain from scratch.
- Faster updates.
- Suitable for real-time applications.

---

# Disadvantages

- Sensitive to poor-quality or noisy data.
- Can forget older patterns if not designed carefully.
- More difficult to implement than batch learning.

---

# Batch Learning vs Online Learning

| Feature | Batch Learning | Online Learning |
|----------|----------------|-----------------|
| Training Data | Entire dataset at once | Data arrives one sample or small batch at a time |
| Learning | Stops after training | Continues after deployment |
| New Data | Requires retraining | Updates automatically |
| Speed | Slower updates | Faster updates |
| Best For | Static datasets | Continuously changing datasets |
| Example | House Price Prediction | Spam Detection |

---

# Everyday Analogy

### Batch Learning

You finish reading an entire textbook before learning anything new.

If the textbook changes, you study the whole book again.

### Online Learning

You read one new page every day and continuously improve your knowledge without starting over.

---

# Key Points to Remember

- **Batch Learning**
  - Trains on the entire dataset.
  - Must be retrained when new data arrives.
  - Best for static datasets.

- **Online Learning**
  - Learns continuously from new data.
  - Updates without retraining from scratch.
  - Best for real-time and changing environments.

---

# Interview Definition

> **Batch learning trains a model on the entire dataset at once and requires retraining when new data arrives, whereas online learning continuously updates the model as new data becomes available without retraining from scratch.**

# ***Instance-Based Learning vs Model-Based Learning***

Machine learning algorithms generally learn in **two different ways**:

1. **Instance-Based Learning**
2. **Model-Based Learning**

The main difference is **how the algorithm makes predictions**.

---

# Instance-Based Learning

## Definition

**Instance-Based Learning** is a learning approach where the model **memorizes the training data** instead of building a general mathematical model. When a new data point arrives, it compares it with the stored training examples and predicts the output based on the most similar instances.

> **Definition (Interview):**
> Instance-based learning is a machine learning approach where the model stores training examples and makes predictions for new data by comparing it with the most similar stored instances.

---

# Simple Explanation

Imagine you want to identify a fruit.

Instead of learning rules like:

- Apples are usually round.
- Apples have stems.

You simply remember pictures of many fruits.

When you see a new fruit, you compare it with the pictures you've already seen and choose the one that looks most similar.

This is **instance-based learning**.

---

# Workflow

```text
Training Data
      │
      ▼
Store Training Examples
      │
      ▼
New Data
      │
      ▼
Compare with Stored Examples
      │
      ▼
Prediction
```

---

# Example 1: House Price Prediction

You have data for many houses.

A new house needs its price predicted.

Instead of creating a mathematical formula, the algorithm finds houses that are **most similar** in:

- Size
- Location
- Number of bedrooms

It predicts the price based on these similar houses.

---

# Example 2: Movie Recommendation

Suppose two users have watched almost the same movies.

If one user enjoys a new movie, the system recommends it to the other user because they are similar.

The recommendation is based on **similar users**, not a mathematical model.

---

# Advantages

- Simple and easy to understand.
- Adapts easily when new training data is added.
- Often performs well when similar examples exist.

---

# Disadvantages

- Requires storing the entire training dataset.
- Prediction can be slow because comparisons are made every time.
- Performance decreases with very large datasets.

---

# Common Algorithm

- **K-Nearest Neighbors (KNN)**

---

# Model-Based Learning

## Definition

**Model-Based Learning** is a learning approach where the algorithm learns a **general mathematical model** from the training data. Instead of storing all training examples, it learns patterns and uses the model to make predictions.

> **Definition (Interview):**
> Model-based learning is a machine learning approach where the algorithm learns a mathematical model from the training data and uses that model to make predictions on new data.

---

# Simple Explanation

Imagine learning the formula:

**Area = Length × Width**

Instead of memorizing the area of every rectangle you've ever seen, you learn the rule once.

Whenever you see a new rectangle, you simply apply the formula.

This is **model-based learning**.

---

# Workflow

```text
Training Data
      │
      ▼
Learn Mathematical Model
      │
      ▼
Trained Model
      │
      ▼
New Data
      │
      ▼
Prediction
```

---

# Example 1: House Price Prediction

A dataset contains thousands of houses and their prices.

The algorithm learns the relationship between:

- House size
- Number of bedrooms
- Location

and the house price.

When a new house appears, it predicts the price using the learned model instead of searching through all previous houses.

---

# Example 2: Spam Email Detection

The model is trained on thousands of labeled emails.

It learns patterns that distinguish spam from legitimate emails.

When a new email arrives, it directly predicts **Spam** or **Not Spam** using the learned model.

---

# Advantages

- Faster predictions.
- Does not need to store all training data.
- Scales well to large datasets.

---

# Disadvantages

- Training may take longer.
- If the model is too simple, it may miss important patterns.
- May need retraining when significant new data becomes available.

---

# Common Algorithms

- Linear Regression
- Logistic Regression
- Decision Trees
- Support Vector Machines (SVM)
- Neural Networks

---

# Instance-Based vs Model-Based Learning

| Feature | Instance-Based Learning | Model-Based Learning |
|----------|-------------------------|----------------------|
| Learning Method | Stores training examples | Learns a mathematical model |
| Prediction | Compare with similar instances | Use learned model |
| Training Time | Low | Higher |
| Prediction Time | Slower | Faster |
| Memory Usage | High | Lower |
| Example Algorithm | K-Nearest Neighbors (KNN) | Linear Regression |

---

# Everyday Analogy

### Instance-Based Learning

You recognize a person's face because you've seen similar faces before.

You compare the new face with the faces stored in your memory.

### Model-Based Learning

You learn general rules about facial features and use those rules to recognize new faces without remembering every face you've ever seen.

---

# Key Points to Remember

- **Instance-Based Learning**
  - Stores training examples.
  - Predicts using similarity.
  - Training is simple, prediction is slower.
  - Example: **K-Nearest Neighbors (KNN)**.

- **Model-Based Learning**
  - Learns a mathematical model.
  - Predicts using the learned model.
  - Training takes longer, prediction is faster.
  - Examples: **Linear Regression, Decision Trees, Neural Networks**.

---

# Interview Definition

> **Instance-based learning stores training examples and predicts outputs by comparing new data with similar instances, whereas model-based learning learns a mathematical model from the training data and uses it to make predictions on new data.**

# **Main Challenges of Machine Learning**

Building a machine learning model is not just about choosing an algorithm. There are many challenges that can prevent a model from performing well.

Let's understand each challenge with simple examples.

---

# 1. Insufficient Quantity of Training Data

## What is it?

Machine learning models learn from examples.

If you give them **very few examples**, they cannot learn the underlying pattern.

### Example

Suppose you want to teach a child what a cat looks like.

You only show this:

```text
🐱 Cat 1
🐱 Cat 2
```

Then show this animal:

```text
🐈
```

The child might not recognize it because they have seen only two cats.

Similarly, if a machine learning model sees only **20 house prices**, it cannot accurately predict the price of the 21st house.

### Solution

- Collect more data.
- Use data augmentation (for images).
- Generate synthetic data when appropriate.

---

# 2. Poor Quality Training Data

## What is it?

Even if you have lots of data, it may contain:

- Missing values
- Wrong values
- Duplicate entries
- Typing mistakes
- Noise

This confuses the model.

### Example

Suppose your dataset is:

| Size | Price |
|------|------|
|1000|200|
|1200|240|
|1500|300|
|1800|360|
|2000|**20** ❌|

The last price should probably be **400**, not **20**.

The model may learn incorrect relationships.

### Real-life Analogy

Imagine learning mathematics from a textbook full of mistakes.

You'll likely learn incorrect concepts.

### Solution

- Clean the data.
- Handle missing values.
- Remove duplicates.
- Detect and fix outliers when appropriate.

---

# 3. Non-Representative Training Data

## What is it?

Your training data should represent the real-world situations where the model will be used.

### Example

Suppose you're building a model to predict house prices across India.

But your training data contains houses only from Mumbai.

The model may perform poorly for cities like Jaipur, Lucknow, or Chennai.

### Real-life Analogy

Imagine learning English by talking only to one person.

You'll understand that person's style, but not everyone else's.

### Solution

Collect data from different regions, users, conditions, or scenarios so the dataset reflects the real world.

---

# 4. Irrelevant Features

## What is it?

Sometimes datasets include information that has nothing to do with the prediction.

### Example

Predicting house prices using:

```text
House Size ✅
Bedrooms ✅
Location ✅
Owner's Favorite Color ❌
Owner's Lucky Number ❌
```

The last two features don't help predict the price.

They may even confuse the model.

### Solution

Use **feature selection** or **feature engineering** to keep only useful information.

---

# 5. Overfitting

## What is it?

The model memorizes the training data instead of learning the general pattern.

It performs very well on the training data but poorly on new, unseen data.

### Example

Suppose a student memorizes every question from last year's exam.

If the teacher asks different questions this year, the student struggles.

The student memorized instead of understanding the concepts.

Machine learning models can do the same.

### Example

Training Accuracy

```text
99%
```

Testing Accuracy

```text
65%
```

The model has likely overfit the training data.

### Solution

- Collect more data.
- Simplify the model.
- Use regularization.
- Use techniques like cross-validation.
- Apply early stopping in neural networks.

---

# 6. Underfitting

## What is it?

The model is too simple to learn the pattern.

It performs poorly on both training and testing data.

### Example

Trying to predict house prices using only:

```text
Price = 100
```

No matter the size or location, the model predicts the same value.

It hasn't learned anything useful.

### Example

Training Accuracy

```text
60%
```

Testing Accuracy

```text
58%
```

Poor performance on both datasets suggests underfitting.

### Solution

- Use a more powerful model.
- Add relevant features.
- Train for longer (when appropriate).
- Reduce excessive regularization.

---

# 7. Data Drift (Changing Data)

## What is it?

The data changes over time.

The relationships the model learned may no longer be valid.

### Example

A house price model trained in **2018** may not accurately predict prices in **2026** because the market has changed.

### Real-life Analogy

Imagine studying road maps from ten years ago.

New roads, flyovers, and traffic patterns may make the old map unreliable.

### Solution

- Retrain the model periodically.
- Monitor model performance.
- Continuously collect fresh data.

---

# 8. Computational Cost

## What is it?

Some models require:

- Large memory
- Powerful GPUs
- Long training times

### Example

Training a deep learning model on millions of images can take hours or even days.

### Solution

- Use better hardware.
- Optimize the model.
- Use cloud computing.
- Reduce unnecessary model complexity.

---

# 9. Choosing the Right Model

## What is it?

Different problems require different algorithms.

### Example

Predicting house prices:

- Linear Regression ✅
- Decision Tree ✅
- Random Forest ✅

Classifying emails as spam:

- Logistic Regression ✅
- Naive Bayes ✅

Choosing the wrong algorithm can lead to poor performance.

### Solution

Try multiple models, compare their performance, and select the one that works best for your problem.

# **Testing and Validating in Machine Learning**

After training a machine learning model, we need to answer an important question:

> **"Will the model perform well on new, unseen data?"**

To answer this, we divide our dataset into different parts and evaluate the model. This process is called **Testing and Validation**.

---

# Why Do We Need Testing and Validation?

Imagine you are preparing for an exam.

If you only solve the questions from your textbook repeatedly, you may memorize them.

But the real exam contains **new questions**.

Similarly, a machine learning model should not just memorize the training data—it should perform well on **new, unseen data**.

---

# Dataset Split

A dataset is usually divided into three parts:

```text
Complete Dataset
       │
       ├───────────────┐
       │               │
 Training Data     Testing Data
       │
       │
 Validation Data (optional)
```

Or more commonly:

```text
Complete Dataset

↓

Training Set (70% - 80%)

↓

Testing Set (20% - 30%)
```

If validation is used:

```text
Training Set : 70%
Validation Set : 15%
Testing Set : 15%
```

---

# 1. Training Set

## What is it?

The **training set** is used to teach the model.

The model learns patterns, relationships, and parameters from this data.

### Example

Suppose you have **1,000 house price records**.

```text
Training Set = 800 houses
```

The model learns:

- Larger houses usually cost more.
- Better locations increase price.
- More bedrooms often increase price.

The model adjusts its parameters using these 800 examples.

---

# 2. Validation Set

## What is it?

The **validation set** is used while developing the model.

It helps us:

- Choose the best algorithm
- Tune hyperparameters
- Detect overfitting

The model **does not learn** from the validation data.

It only uses it for evaluation during training.

### Example

Suppose you are deciding between:

- Linear Regression
- Decision Tree
- Random Forest

You train all three models using the training set.

Then evaluate them on the **validation set**.

Suppose the results are:

| Model | Validation Accuracy |
|--------|---------------------|
| Linear Regression | 82% |
| Decision Tree | 87% |
| Random Forest | 91% ✅ |

You choose the Random Forest model because it performs the best on the validation data.

---

# 3. Testing Set

## What is it?

The **testing set** is used only once, after the final model has been selected.

It tells us how well the model performs on completely unseen data.

### Example

Suppose your final model is Random Forest.

You now test it using the **200 house records** that it has never seen before.

Suppose:

```text
Training Accuracy = 95%

Testing Accuracy = 93%
```

Since the testing accuracy is close to the training accuracy, the model generalizes well.

---

# Complete Example

Suppose we have **1,000 house records**.

```text
1000 Houses

↓

700 → Training

150 → Validation

150 → Testing
```

### Step 1

Train the model using:

```text
700 houses
```

---

### Step 2

Evaluate using:

```text
150 validation houses
```

Choose the best algorithm and tune hyperparameters.

---

### Step 3

Finally,

Evaluate using:

```text
150 testing houses
```

This gives the final estimate of real-world performance.

---

# Why Can't We Test on the Training Data?

Imagine a student who memorizes last year's exam paper.

Teacher asks the exact same questions.

The student scores:

```text
100%
```

But on a new exam:

```text
60%
```

The student memorized instead of learning.

Machine learning models behave the same way.

Testing on the training data gives an overly optimistic result.

---

# Overfitting Example

Suppose

```text
Training Accuracy = 99%

Testing Accuracy = 68%
```

The model has memorized the training data.

This is called **Overfitting**.

---

# Good Model Example

```text
Training Accuracy = 94%

Testing Accuracy = 92%
```

The model performs similarly on both datasets.

This indicates good generalization.

---

# Underfitting Example

```text
Training Accuracy = 60%

Testing Accuracy = 58%
```

The model performs poorly everywhere.

This indicates **Underfitting**.

---

# Cross-Validation

Sometimes, instead of using one validation set, we use **Cross-Validation**.

The most common method is **K-Fold Cross-Validation**.

Suppose we have:

```text
1000 records
```

Choose:

```text
K = 5
```

The dataset is divided into:

```text
Fold 1
Fold 2
Fold 3
Fold 4
Fold 5
```

The model trains on **4 folds** and validates on **1 fold**.

This process repeats five times.

```text
Iteration 1

Train → Fold 2,3,4,5

Validate → Fold 1

-------------------------

Iteration 2

Train → Fold 1,3,4,5

Validate → Fold 2

-------------------------

Iteration 3

Train → Fold 1,2,4,5

Validate → Fold 3

...
```

Finally,

Average the validation scores.

Cross-validation gives a more reliable estimate because every record gets used for both training and validation.
